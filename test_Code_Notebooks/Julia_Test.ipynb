```julia
function sayhi(name)
    println("hi $name")
end
sayhi("Tim")
```

```julia
function f(x)
    x^2
end
f(5)
```

```julia
# Can also create funciotn in a single line
sayhi2(name) = println("hi $name")
f2(x) = x^2
```

```julia
# Can also define them as maps.
sayhi3 = name -> println("hi $name")
f3 = x -> x^2
```
